```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the cleaned dataset (using bezdekIris since it's the corrected version)
columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']
df = pd.read_csv('bezdekIris.data', names=columns, header=None)
```

```python
# 1. Basic descriptive statistics
print("--- Class Distribution ---")
print(df['class'].value_counts())

print("\n--- Summary Statistics by Class ---")
print(df.groupby('class').mean())
```

```python
# 2. Pairplot to visualize relationships between all pairs of features
sns.set_theme(style="ticks")
pair_plot = sns.pairplot(df, hue="class", markers=["o", "s", "D"])
pair_plot.savefig('iris_pairplot.png')
plt.show()
```

```python
# 3. Correlation Heatmap (only numeric columns)
fig, ax = plt.subplots(figsize=(8, 6))
numeric_df = df.drop(columns=['class'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f", ax=ax)
ax.set_title('Correlation Matrix of Iris Features')
plt.tight_layout()
plt.savefig('iris_correlation_heatmap.png')
plt.show()
```

```python
# 4. Boxplots to see distributions and outliers
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()
for i, col in enumerate(columns[:4]):
    sns.boxplot(data=df, x='class', y=col, ax=axes[i], palette='Set2')
    axes[i].set_title(f'Distribution of {col.replace("_", " ").title()}')
    axes[i].set_xlabel('Species')
    axes[i].set_ylabel('Length/Width (cm)')

plt.tight_layout()
plt.savefig('iris_boxplots.png')
plt.show()
```
