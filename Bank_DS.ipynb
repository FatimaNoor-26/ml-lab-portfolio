```python
import pandas as pd

# Define the file paths (assuming they are in your current working directory)
file_paths = {
    "bank_additional": "bank-additional.csv",
    "bank_additional_full": "bank-additional-full.csv",
    "bank": "bank.csv",
    "bank_full": "bank-full.csv"
}
```

```python
# Dictionary to hold the loaded DataFrames
dataframes = {}

# Load each file into a separate DataFrame
for key, path in file_paths.items():
    try:
        # Note: sep=';' is crucial because these files use semicolons as delimiters
        dataframes[key] = pd.read_csv(path, sep=';')
        print(f"Successfully loaded {path} -> Shape: {dataframes[key].shape}")
    except FileNotFoundError:
        print(f"Error: The file {path} was not found. Check the path.")
    except Exception as e:
        print(f"An error occurred while loading {path}: {e}")

# --- Example: Accessing a specific DataFrame ---
# To look at the first few rows of 'bank-additional.csv':
# print(dataframes["bank_additional"].head())
```
