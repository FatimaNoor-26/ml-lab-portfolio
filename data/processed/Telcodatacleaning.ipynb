```python
import pandas as pd
import numpy as np

# 1. Load the dataset
df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')

# 2. Fix 'TotalCharges' column
# Some rows have a space " " instead of a number. We convert these to 0 
# because they correspond to customers with 0 tenure.
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce').fillna(0)

# 3. Handle Categorical Redundancy
# Many columns have "No internet service" or "No phone service" which essentially means "No"
replace_cols = [
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]
for col in replace_cols:
    df[col] = df[col].replace('No internet service', 'No')

df['MultipleLines'] = df['MultipleLines'].replace('No phone service', 'No')

# 4. Binary Encoding (Yes/No and Gender)
# Convert 'Yes' to 1 and 'No' to 0
binary_cols = [
    'Partner', 'Dependents', 'PhoneService', 'MultipleLines', 
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 
    'StreamingTV', 'StreamingMovies', 'PaperlessBilling', 'Churn'
]
for col in binary_cols:
    df[col] = df[col].map({'Yes': 1, 'No': 0})

# Map gender
df['gender'] = df['gender'].map({'Female': 1, 'Male': 0})

# 5. One-Hot Encoding for multi-class columns
# This converts columns like 'Contract' and 'PaymentMethod' into separate binary columns
df = pd.get_dummies(df, columns=['InternetService', 'Contract', 'PaymentMethod'], dtype=int)

# 6. Drop redundant columns
# customerID is just an identifier and won't help the ML model
df.drop('customerID', axis=1, inplace=True)

# 7. Final Check
print("Dataset Cleaned. Info:")
print(df.info())
print("\nFirst 5 rows:")
print(df.head())

# Save the cleaned dataset
df.to_csv('cleaned_telco_churn.csv', index=False)
```
