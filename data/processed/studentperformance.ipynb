```python
import pandas as pd

# 1. Load the dataset
df = pd.read_csv('StudentsPerformance.csv')

# 2. Rename columns to snake_case
# This removes spaces and slashes for easier variable access (e.g., df.math_score)
df.columns = [col.replace(' ', '_').replace('/', '_') for col in df.columns]
df.rename(columns={'parental_level_of_education': 'parental_education'}, inplace=True)

# 3. Feature Engineering: Create Summary Scores
# Adding a total and average score helps capture overall student performance
df['total_score'] = df['math_score'] + df['reading_score'] + df['writing_score']
df['average_score'] = df['total_score'] / 3

# 4. Categorical Encoding
# Binary mapping (0 and 1)
df['gender'] = df['gender'].map({'female': 1, 'male': 0})
df['lunch'] = df['lunch'].map({'standard': 1, 'free/reduced': 0})
df['test_preparation_course'] = df['test_preparation_course'].map({'completed': 1, 'none': 0})

# Ordinal mapping for Education
# We use ordinal mapping because there is a logical order to education levels
edu_map = {
    'some high school': 0,
    'high school': 1,
    'some college': 2,
    "associate's degree": 3,
    "bachelor's degree": 4,
    "master's degree": 5
}
df['parental_education'] = df['parental_education'].map(edu_map)

# One-hot encoding for race/ethnicity
# Since there is no natural order between groups, we create separate dummy columns
df = pd.get_dummies(df, columns=['race_ethnicity'], prefix='race', dtype=int)

# 5. Review Cleaned Data
print("Cleaned Data Preview:")
print(df.head())

# Save the cleaned dataset for your project
df.to_csv('cleaned_student_performance.csv', index=False)
```
