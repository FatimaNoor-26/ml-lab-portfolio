```python
import pandas as pd
import numpy as np

# 1. Load the dataset
df = pd.read_csv('heart_cleveland_upload.csv')

# 2. Rename columns for better readability
# This makes it easier to understand the features during EDA
df.columns = [
    'age', 'sex', 'chest_pain', 'resting_blood_pressure', 
    'cholesterol', 'fasting_blood_sugar', 'rest_ecg', 
    'max_heart_rate', 'exercise_angina', 'st_depression', 
    'st_slope', 'num_major_vessels', 'thalassemia', 'condition'
]

# 3. Check for duplicates
if df.duplicated().sum() > 0:
    df.drop_duplicates(inplace=True)
    print(f"Removed {df.duplicated().sum()} duplicate rows.")

# 4. Handle Outliers (Optional but recommended for Heart Data)
# Variables like Cholesterol and Blood Pressure can have extreme outliers 
# that skew linear models. Here we cap them using the IQR method.
continuous_cols = ['resting_blood_pressure', 'cholesterol', 'max_heart_rate', 'st_depression']

for col in continuous_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    # Capping values instead of dropping to preserve data size
    df[col] = np.where(df[col] > upper_bound, upper_bound, df[col])
    df[col] = np.where(df[col] < lower_bound, lower_bound, df[col])

# 5. One-Hot Encoding for Categorical Features
# Features like chest_pain, st_slope, and thalassemia are categorical 
# but represented as integers. ML models might wrongly assume a mathematical order.
categorical_cols = ['chest_pain', 'rest_ecg', 'st_slope', 'thalassemia']
df = pd.get_dummies(df, columns=categorical_cols, drop_first=True, dtype=int)

# 6. Final Data Check
print("Cleaning complete. New Shape:", df.shape)
print("\nMissing values check:")
print(df.isnull().sum())

# 7. Save the cleaned dataset
df.to_csv('cleaned_heart_disease.csv', index=False)
```
