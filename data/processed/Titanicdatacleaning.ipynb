```python
import pandas as pd

# 1. Load the dataset
df = pd.read_csv('train.csv')

# 2. Handle Missing Values
# Fill missing Age with the median value (robust to outliers)
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing Embarked with the most common port (mode)
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# 3. Drop Unnecessary Columns
# Cabin has ~77% missing data; PassengerId, Name, and Ticket are unique identifiers 
# that don't contribute to general patterns.
df.drop(columns=['Cabin', 'PassengerId', 'Name', 'Ticket'], inplace=True)

# 4. Encode Categorical Variables
# Convert Sex to binary (0 for male, 1 for female)
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

# One-Hot Encode the Embarked column (creates columns for C, Q, and S)
df = pd.get_dummies(df, columns=['Embarked'], prefix='Embarked', dtype=int)

# 5. Review Cleaned Data
print("Data Cleaning Complete. Missing values:")
print(df.isnull().sum())
print("\nFirst 5 rows of cleaned data:")
print(df.head())

# Save the cleaned data to a new file
df.to_csv('cleaned_titanic.csv', index=False)
```
